# AMO Web Application - API Beispiele

Dieses Notebook zeigt, wie Sie die AMO Web-API programmgesteuert verwenden können, anstatt das Web-Interface zu nutzen.

**Voraussetzung:** Der Web-Server muss laufen (`python webapp.py` oder `start_webapp.ps1`)

```python
import requests
import json
from pathlib import Path
```

## 1. Verfügbare Modelle abrufen

Rufen Sie die Liste der verfügbaren Orchestrierungsmodelle ab:

```python
# API Basis-URL
BASE_URL = "http://localhost:8001"

# Verfügbare Modelle abrufen
response = requests.get(f"{BASE_URL}/api/models")
models = response.json()

print("Verfügbare Modelle:")
print("-" * 50)
for model in models:
    status = "✓ Verfügbar" if model['available'] else "✗ Nicht verfügbar"
    print(f"{model['name']}: {status}")
    print(f"  {model['description']}")
    print()
```

## 2. MIDI-Datei orchestrieren

Laden Sie eine MIDI-Datei hoch und orchestrieren Sie sie mit einem ausgewählten Modell:

```python
# Pfad zur MIDI-Datei
midi_file = "data/samples/midis/fur-elise.mid"

# Modell auswählen (beethoven, debussy, oder tchaikovsky)
selected_model = "beethoven"

# Datei hochladen und orchestrieren
with open(midi_file, 'rb') as f:
    files = {'file': (Path(midi_file).name, f, 'audio/midi')}
    data = {'model': selected_model}
    
    response = requests.post(
        f"{BASE_URL}/api/orchestrate",
        files=files,
        data=data
    )

# Ergebnis anzeigen
if response.status_code == 200:
    result = response.json()
    print("✓ Orchestrierung erfolgreich!")
    print(f"Modell verwendet: {result['model_used']}")
    print(f"Ausgabedatei: {result['output_file']}")
    output_filename = result['output_file']
else:
    print(f"✗ Fehler: {response.status_code}")
    print(response.json())
```

## 3. Orchestrierte Datei herunterladen

Laden Sie die orchestrierte MIDI-Datei herunter:

```python
# Datei herunterladen (verwenden Sie den output_filename aus dem vorherigen Schritt)
download_url = f"{BASE_URL}/api/download/{output_filename}"
response = requests.get(download_url)

if response.status_code == 200:
    # Speichern Sie die Datei lokal
    save_path = f"downloaded_{output_filename}"
    with open(save_path, 'wb') as f:
        f.write(response.content)
    print(f"✓ Datei heruntergeladen: {save_path}")
else:
    print(f"✗ Download fehlgeschlagen: {response.status_code}")
```

## 4. Komplettes Beispiel: Alle Modelle testen

Orchestrieren Sie eine MIDI-Datei mit allen verfügbaren Modellen:

```python
def orchestrate_with_all_models(midi_file_path):
    """
    Orchestriert eine MIDI-Datei mit allen verfügbaren Modellen.
    """
    # Modelle abrufen
    response = requests.get(f"{BASE_URL}/api/models")
    models = response.json()
    
    results = []
    
    for model in models:
        if not model['available']:
            print(f"⊘ {model['name']}: Nicht verfügbar, überspringe...")
            continue
        
        print(f"Orchestriere mit {model['name']}...")
        
        try:
            # Datei hochladen und orchestrieren
            with open(midi_file_path, 'rb') as f:
                files = {'file': (Path(midi_file_path).name, f, 'audio/midi')}
                data = {'model': model['id']}
                
                response = requests.post(
                    f"{BASE_URL}/api/orchestrate",
                    files=files,
                    data=data
                )
            
            if response.status_code == 200:
                result = response.json()
                print(f"  ✓ Erfolgreich: {result['output_file']}")
                
                # Datei herunterladen
                download_url = f"{BASE_URL}/api/download/{result['output_file']}"
                dl_response = requests.get(download_url)
                
                if dl_response.status_code == 200:
                    save_path = f"output_{result['output_file']}"
                    with open(save_path, 'wb') as f:
                        f.write(dl_response.content)
                    print(f"  ✓ Gespeichert: {save_path}")
                    results.append({
                        'model': model['name'],
                        'file': save_path,
                        'success': True
                    })
                else:
                    print(f"  ✗ Download fehlgeschlagen")
                    results.append({
                        'model': model['name'],
                        'success': False,
                        'error': 'Download failed'
                    })
            else:
                error = response.json()
                print(f"  ✗ Orchestrierung fehlgeschlagen: {error.get('detail', 'Unknown error')}")
                results.append({
                    'model': model['name'],
                    'success': False,
                    'error': error.get('detail', 'Unknown error')
                })
        
        except Exception as e:
            print(f"  ✗ Fehler: {str(e)}")
            results.append({
                'model': model['name'],
                'success': False,
                'error': str(e)
            })
        
        print()
    
    return results

# Beispiel verwenden
results = orchestrate_with_all_models("data/samples/midis/fur-elise.mid")

# Zusammenfassung
print("\n" + "=" * 60)
print("ZUSAMMENFASSUNG")
print("=" * 60)
for result in results:
    status = "✓" if result['success'] else "✗"
    print(f"{status} {result['model']}")
    if result['success']:
        print(f"   → {result['file']}")
    else:
        print(f"   → Fehler: {result.get('error', 'Unknown')}")
```

## 5. Server-Gesundheitsprüfung

Überprüfen Sie, ob der Server läuft und wie viele Modelle verfügbar sind:

```python
try:
    response = requests.get(f"{BASE_URL}/api/health", timeout=2)
    if response.status_code == 200:
        health = response.json()
        print("✓ Server läuft")
        print(f"  Status: {health['status']}")
        print(f"  Verfügbare Modelle: {health['available_models']}")
    else:
        print("✗ Server antwortet mit Fehler")
except requests.exceptions.RequestException as e:
    print("✗ Server nicht erreichbar")
    print(f"  Stellen Sie sicher, dass der Server läuft: python webapp.py")
    print(f"  Fehler: {e}")
```

## Tipps

- **Server starten:** Führen Sie `python webapp.py` oder `start_webapp.ps1` in einem separaten Terminal aus
- **Port ändern:** Wenn Port 8000 belegt ist, ändern Sie in `webapp.py` den Port und passen Sie `BASE_URL` hier an
- **Fehlerbehandlung:** Überprüfen Sie immer den Response-Status-Code und behandeln Sie Fehler entsprechend
- **Große Dateien:** Für sehr große MIDI-Dateien kann die Orchestrierung einige Zeit dauern
